```python
import numpy as np

PLAYER = 3
SEASON = 1
GP = 5
MIN = 6
FGM = 7
FGA = 8
TPM = 9
TPA = 10
FTM = 11
FTA = 12
STL = 18
BLK = 19
PTS = 20

class NBAStats:
    def __init__(self, file_path):
        #Load text data (player names, seasons, etc.)
        self.text = np.genfromtxt(
            file_path,
            delimiter="\t",
            dtype=str,
            skip_header=1
        )

        #Load numeric data (stats only)
        #Missing values are automatically replaced with 0
        self.num = np.genfromtxt(
            file_path,
            delimiter="\t",
            skip_header=1,
            filling_values=0
        )

    def safe_divide(self, top, bottom):
        return np.divide(top, bottom, out=np.zeros(len(top)), where=bottom != 0)

    def calculate_metrics(self):
        #Shooting accuracies
        self.fg_acc = self.safe_divide(self.num[:, FGM], self.num[:, FGA])
        self.tp_acc = self.safe_divide(self.num[:, TPM], self.num[:, TPA])
        self.ft_acc = self.safe_divide(self.num[:, FTM], self.num[:, FTA])

        #Overall shooting accuracy
        self.overall_acc = self.safe_divide(
            self.num[:, FGM] + self.num[:, FTM],
            self.num[:, FGA] + self.num[:, FTA]
        )

        #Scoring metrics
        self.ppm = self.safe_divide(self.num[:, PTS], self.num[:, MIN])
        self.ppg = self.safe_divide(self.num[:, PTS], self.num[:, GP])

        #Defensive metrics
        self.bpg = self.safe_divide(self.num[:, BLK], self.num[:, GP])
        self.spg = self.safe_divide(self.num[:, STL], self.num[:, GP])

        #Combine all results into one ndarray
        self.results = np.column_stack((
            self.text[:, PLAYER],   #Player name
            self.text[:, SEASON],   #Season
            self.fg_acc,
            self.tp_acc,
            self.ft_acc,
            self.ppm,
            self.overall_acc,
            self.bpg,
            self.spg,
            self.ppg
        ))

    def top_100(self, metric_index, label):
        """
        Sorts the results array by the selected metric
        and prints the top 100 player-seasons.
        """
        sorted_rows = self.results[
            self.results[:, metric_index].astype(float).argsort()[::-1]
        ]

        print("\nTop 100 for", label)
        for row in sorted_rows[:100]:
            print(row[0], row[1], row[metric_index])
            
#Program Execution

#Create NBAStats object
stats = NBAStats("NBA_Player_Stats.tsv")

#Calculate all metrics
stats.calculate_metrics()

#Display rankings
stats.top_100(2, "Field Goal Accuracy")
stats.top_100(3, "Three Point Accuracy")
stats.top_100(4, "Free Throw Accuracy")
stats.top_100(9, "Points Per Game")
stats.top_100(6, "Overall Shooting Accuracy")
stats.top_100(7, "Blocks Per Game")
stats.top_100(8, "Steals Per Game")
```
